**RESPOSTAS**


1. Um algoritmo é uma sequência finita de passos bem definidos e organizados que são executados para resolver um problema ou realizar uma tarefa.



2. *   Linguagem de alto nível: É mais próxima da linguagem humana, mais fácil de aprender e entender, como Python, Java, C#. Tem comandos mais intuitivos.
*   Linguagem de baixo nível: É mais próxima da linguagem de máquina. Exige conhecimento mais técnico, como Assembly. É mais difícil de entender, mas permite maior controle do hardware.



3. Lógica é a ciência do raciocínio. Na programação, a lógica é usada para construir soluções corretas e eficientes, com base em sequências, decisões e repetições.

4. * Entrada: São os dados fornecidos ao algoritmo (ex: notas, nomes, salários).

* Saída: São os resultados que o algoritmo produz (ex: média, situação, novo salário).

* Relação: O algoritmo transforma as entradas (dados iniciais) em saídas (resultados) por meio de um conjunto de instruções.



5. * Análise do problema: Entender claramente o que o programa deve resolver.

* Planejamento (algoritmo): Criar um passo a passo lógico da solução.

* Codificação: Escrever o algoritmo em uma linguagem de programação.

* Compilação/interpretação: Traduzir o código para linguagem de máquina.

* Testes e depuração: Verificar e corrigir erros.

* Execução: Rodar o programa para resolver o problema real.

* Manutenção: Fazer melhorias ou correções com o tempo.

````python
#  Questão 3 - Troca de valores A e B

a = input("Digite o valor de A: ")
b = input("Digite o valor de B: ")

a, b = b, a

print("Novo valor de A:", a)
print("Novo valor de B:", b)
````

````python
Algoritmo TrocaValores
   Declare a, b, temp: Caracter

   Escreva("Digite o valor de A: ")
   Leia(a)
   Escreva("Digite o valor de B: ")
   Leia(b)

   temp <- a
   a <- b
   b <- temp

   Escreva("Novo valor de A: ", a)
   Escreva("Novo valor de B: ", b)
Fim
````

````python
#  Questão 4 - Média de 3 notas de um aluno

nome = input("Digite o nome do aluno: ")
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = (nota1 + nota2 + nota3) / 3
print("Aluno:", nome)
print("Média:", media)
````

````python
#  Questão 1 – Peso ideal baseado na altura e sexo

sexo = input("Digite o sexo (M para masculino, F para feminino): ").upper()
altura = float(input("Digite a altura (em metros): "))

if sexo == "M":
    peso_ideal = (72.7 * altura) - 58
elif sexo == "F":
    peso_ideal = (62.1 * altura) - 44.7
else:
    peso_ideal = None
    print("Sexo inválido.")

if peso_ideal is not None:
    print("Peso ideal:", round(peso_ideal, 2), "kg")
````

````python
Algoritmo PesoIdeal
   Declare sexo: Caracter
   Declare altura, peso_ideal: Real

   Escreva("Digite o sexo (M/F): ")
   Leia(sexo)
   Escreva("Digite a altura: ")
   Leia(altura)

   Se sexo = "M" Então
      peso_ideal <- (72.7 * altura) - 58
   Senão
      Se sexo = "F" Então
         peso_ideal <- (62.1 * altura) - 44.7
      Senão
         Escreva("Sexo inválido.")
         FimAlgoritmo
      FimSe
   FimSe

   Escreva("Peso ideal: ", peso_ideal)
Fim
````

````python
#   Questão 2 – Categoria de natação por idade

idade = int(input("Digite a idade do nadador: "))

if idade <= 8:
    categoria = "Infantil A"
elif idade < 13:
    categoria = "Infantil B"
elif idade < 18:
    categoria = "Juvenil A"
elif idade < 21:
    categoria = "Juvenil B"
else:
    categoria = "Sênior"

print("Categoria:", categoria)
````

````python
#  Questão 3 – Comparação entre duas datas

print("Digite a primeira data:")
dia1 = int(input("Dia: "))
mes1 = int(input("Mês: "))
ano1 = int(input("Ano: "))

print("Digite a segunda data:")
dia2 = int(input("Dia: "))
mes2 = int(input("Mês: "))
ano2 = int(input("Ano: "))

if (ano1, mes1, dia1) > (ano2, mes2, dia2):
    print("A primeira data é mais recente.")
elif (ano1, mes1, dia1) < (ano2, mes2, dia2):
    print("A segunda data é mais recente.")
else:
    print("As datas são iguais.")
````

````python
Algoritmo CompararDatas
   Declare dia1, mes1, ano1, dia2, mes2, ano2: Inteiro

   Escreva("Digite a primeira data:")
   Leia(dia1)
   Leia(mes1)
   Leia(ano1)

   Escreva("Digite a segunda data:")
   Leia(dia2)
   Leia(mes2)
   Leia(ano2)

   Se ano1 > ano2 ou (ano1 = ano2 e mes1 > mes2) ou (ano1 = ano2 e mes1 = mes2 e dia1 > dia2) Então
      Escreva("A primeira data é mais recente.")
   Senão Se ano1 < ano2 ou (ano1 = ano2 e mes1 < mes2) ou (ano1 = ano2 e mes1 = mes2 e dia1 < dia2) Então
      Escreva("A segunda data é mais recente.")
   Senão
      Escreva("As datas são iguais.")
   FimSe
Fim
````

````python
#  Questão 4 – Verificar se o ano é bissexto

ano = int(input("Digite um ano: "))

if (ano % 400 == 0) or (ano % 4 == 0 and ano % 100 != 0):
    print("O ano é bissexto.")
else:
    print("O ano não é bissexto.")
````

````python
Algoritmo AnoBissexto
   Declare ano: Inteiro

   Escreva("Digite um ano: ")
   Leia(ano)

   Se (ano mod 400 = 0) ou (ano mod 4 = 0 e ano mod 100 <> 0) Então
      Escreva("O ano é bissexto.")
   Senão
      Escreva("O ano não é bissexto.")
   FimSe
Fim
````

````python
#  Questão 5 – Desconto baseado no valor da compra

valor = float(input("Digite o valor da compra: "))

if valor >= 500:
    desconto = valor * 0.25
elif valor >= 200:
    desconto = valor * 0.20
else:
    desconto = valor * 0.15

total = valor - desconto

print("Valor da compra:", valor)
print("Desconto:", desconto)
print("Total a pagar:", total)
````

````python
Algoritmo DescontoCompra
   Declare valor, desconto, total: Real

   Escreva("Digite o valor da compra: ")
   Leia(valor)

   Se valor >= 500 Então
      desconto <- valor * 0.25
   Senão Se valor >= 200 Então
      desconto <- valor * 0.20
   Senão
      desconto <- valor * 0.15
   FimSe

   total <- valor - desconto

   Escreva("Valor da compra: ", valor)
   Escreva("Desconto: ", desconto)
   Escreva("Total a pagar: ", total)
Fim
````

````python
#  Questão 6 – Correção de múltipla escolha

print("Digite suas respostas (a, b, c ou d):")
res1 = input("Questão 1: ").lower()
res2 = input("Questão 2: ").lower()
res3 = input("Questão 3: ").lower()

print("Digite o gabarito:")
gab1 = input("Gabarito 1: ").lower()
gab2 = input("Gabarito 2: ").lower()
gab3 = input("Gabarito 3: ").lower()

acertos = 0
if res1 == gab1:
    acertos += 1
if res2 == gab2:
    acertos += 1
if res3 == gab3:
    acertos += 1

print("Total de acertos:", acertos)
````

````python
Algoritmo CorrecaoQuestoes
   Declare res1, res2, res3, gab1, gab2, gab3: Caracter
   Declare acertos: Inteiro

   acertos <- 0

   Escreva("Digite suas respostas:")
   Leia(res1)
   Leia(res2)
   Leia(res3)

   Escreva("Digite o gabarito:")
   Leia(gab1)
   Leia(gab2)
   Leia(gab3)

   Se res1 = gab1 Então
      acertos <- acertos + 1
   FimSe
   Se res2 = gab2 Então
      acertos <- acertos + 1
   FimSe
   Se res3 = gab3 Então
      acertos <- acertos + 1
   FimSe

   Escreva("Total de acertos: ", acertos)
Fim
````

````python
# Questão 1 - Controle de Estoque Específico
codigo_x = int(input("Informe o código do produto a verificar: "))
total_vendido = 0

while True:
    codigo = int(input("Digite o código do produto (0 para parar): "))
    if codigo == 0:
        break
    quantidade = int(input("Digite a quantidade vendida: "))
    if codigo == codigo_x:
        total_vendido += quantidade

print("Total vendido do produto", codigo_x, ":", total_vendido)
````

````python
# Questão 2 - Processamento de Notas da Turma
n = int(input("Digite o número de alunos: "))
aprovados = 0
reprovados = 0

for i in range(n):
    print(f"Aluno {i+1}:")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))
    media = (nota1 + nota2 + nota3) / 3
    if media >= 7.0:
        print("Aprovado")
        aprovados += 1
    else:
        print("Reprovado")
        reprovados += 1

print("Total aprovados:", aprovados)
print("Total reprovados:", reprovados)
````

````python
# Questão 3 - Simulador de Caixa Eletrônico
saldo = float(input("Informe o saldo inicial: "))

while True:
    operacao = input("Digite a operação (D/S/F): ").upper()
    if operacao == 'F':
        break
    valor = float(input("Digite o valor: "))
    if operacao == 'D':
        saldo += valor
    elif operacao == 'S':
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print("Saldo final:", saldo)
````

````python
# Questão 4 - Inspeção de Qualidade
n = int(input("Digite o número de peças: "))
boas = 0

for i in range(n):
    medida = float(input(f"Medida da peça {i+1}: "))
    if 9.5 <= medida <= 10.5:
        boas += 1

print("Peças dentro do padrão:", boas)
````

````python
# Questão 5 - Análise Pluviométrica
total_chuva = 0
maior = 0
dia_maior = 0

for dia in range(1, 31):
    chuva = float(input(f"Chuva no dia {dia}: "))
    total_chuva += chuva
    if chuva > maior:
        maior = chuva
        dia_maior = dia

media_chuva = total_chuva / 30
print("Total de chuva:", total_chuva)
print("Média de chuva:", media_chuva)
print("Dia com mais chuva:", dia_maior)
````

````python
# Questão 6 - Simulação de Investimento
saldo = 500
for mes in range(12):
    saldo += 100  # aporte
    saldo *= 1.005  # juros de 0.5%

print("Saldo após 12 meses:", round(saldo, 2))
````

````python
# Questão 7 - Pesquisa de Satisfação
total_idade = 0
total_nota = 0
count = 0
promotores = 0

while True:
    idade = int(input("Digite a idade (0 para parar): "))
    if idade == 0:
        break
    nota = int(input("Nota de satisfação (1 a 5): "))
    total_idade += idade
    total_nota += nota
    count += 1
    if nota >= 4:
        promotores += 1

if count > 0:
    print("Idade média:", total_idade / count)
    print("Nota média:", total_nota / count)
    print("Promotores:", promotores)
else:
    print("Nenhum dado registrado.")
````

````python
# Questão 8 - Atividade Física
passos_totais = 0
meta_dias = 0

for dia in range(7):
    passos = int(input(f"Passos no dia {dia+1}: "))
    passos_totais += passos
    if passos >= 10000:
        meta_dias += 1

media_passos = passos_totais / 7
print("Passos totais:", passos_totais)
print("Média diária:", media_passos)
print("Dias com meta atingida:", meta_dias)
````

````python
# Questão 9 - Caixa Registradora
total = 0

while True:
    preco = float(input("Digite o preço do item (0 para parar): "))
    if preco == 0:
        break
    total += preco

if total > 200:
    total *= 0.9  # aplica 10% de desconto

print("Valor a pagar:", round(total, 2))
````

````python
# Questão 10 - Crescimento Populacional
pop = int(input("População inicial: "))
taxa = float(input("Taxa de crescimento (%): "))
pop_alvo = int(input("População alvo: "))
anos = 0

while pop < pop_alvo:
    pop *= (1 + taxa / 100)
    anos += 1

print("Anos necessários:", anos)
````

````python
# Questão 11 - Conta de Energia Média
meses = int(input("Quantos meses: "))
total_conta = 0

for i in range(meses):
    consumo = float(input(f"Consumo no mês {i+1} (kWh): "))
    if consumo <= 100:
        custo = consumo * 0.5
    else:
        custo = 100 * 0.5 + (consumo - 100) * 0.75
    custo += 10  # taxa fixa
    total_conta += custo

media = total_conta / meses
print("Média da conta de energia:", round(media, 2))
````
